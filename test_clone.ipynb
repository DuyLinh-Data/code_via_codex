```python
# Convert column headers to lowercase

import pandas as pd
import re

df.columns = [re.sub(r'\W+', '_', col.lower().strip()) for col in df.columns]
```

```python
# Check datatype
print(df['date'].dtypes)  
```

```python
# Convert to datetime
df['date'] = pd.to_datetime(df['date'])
```

```python
# Check datatype
print(df['date'].dtypes)  # Output: object
```

```python
# Check for non-integer values in numeric columns
non_integers = df.select_dtypes(include=['float64']).applymap(lambda x: x != int(x) if pd.notna(x) else False).sum()
print("Non-integer counts by column:")
print(non_integers)
print("\nColumns with non-integer values:")
print(non_integers[non_integers > 0])
```
